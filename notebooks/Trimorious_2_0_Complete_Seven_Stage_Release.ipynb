### 05 — Frozen D07 scoring rules
This cell writes the implementation; the final controller executes it when prerequisites pass.

````python
%%writefile /tmp/trimorious_v2_auto/scorer.py
if STAGE=='EVALUATE':
    def options(t): return dict(re.findall(r'^([ABCD])\. (.+)$',t['question'],re.M))

    def last_object(text):
        # Parse final JSON object (including nested tool args); never eval model output.
        text=str(text).strip()
        if len(text)>20000: return None
        def reject_nonfinite(s): raise ValueError('Non-finite JSON constant')
        dec=json.JSONDecoder(parse_constant=reject_nonfinite); candidates=[]
        for m in re.finditer(r'\{',text):
            try:
                obj,end=dec.raw_decode(text[m.start():]); candidates.append((m.start()+end,m.start(),obj))
            except (ValueError,RecursionError): pass
        if not candidates: return None
        _,_,obj=max(candidates,key=lambda x:(x[0],-x[1]))
        if not isinstance(obj,dict): return None
        try: canon(obj)
        except (ValueError,TypeError,OverflowError): return None
        return obj

    def numeric_ok(x,gold,atol,rtol):
        if isinstance(x,bool) or not isinstance(x,(int,float)): return False
        try: return math.isfinite(float(x)) and math.isclose(float(x),gold,abs_tol=atol,rel_tol=rtol)
        except (OverflowError,ValueError): return False

    def execute_tool(task, raw):
        obj=last_object(raw)
        if not obj or set(obj)!={'tool','args'} or obj['tool']!=task['tool'] or canon(obj['args'])!=canon(task['tool_args']):
            return {'ok':False,'error':'Invalid or incorrect tool call; no tool executed.'}
        args=obj['args']; name=obj['tool']
        # Exact argument matching limits execution to the small authored fixture for each task.
        if name=='mean': val=statistics.mean(args['values'])
        elif name=='sample_variance': val=statistics.variance(args['values'])
        elif name=='dot': val=sum(x*y for x,y in zip(args['a'],args['b']))
        else: return {'ok':False,'error':'Unknown tool'}
        return {'ok':True,'value':val}

    def score(task, raw, tool_result=None):
        obj=last_object(raw); protocol=bool(obj is not None and set(obj)=={'answer'})
        val=obj.get('answer') if protocol else None
        kind=task['kind']; correct=None
        if kind=='rubric': protocol=bool(protocol and isinstance(val,str) and val.strip())
        if kind=='choice': correct=bool(protocol and isinstance(val,str) and val.strip().upper()==task['gold'])
        elif kind in ('numeric','tool_numeric'):
            correct=bool(protocol and numeric_ok(val,task['gold'],task['atol'],task['rtol']))
            if kind=='tool_numeric': correct=bool(correct and tool_result and tool_result.get('ok'))
        return {'score':None if correct is None else int(correct),'protocol_ok':protocol,'parsed_answer':val,
                'review_status':'human_review_required' if kind=='rubric' else 'objective_scored'}

    def prompt_for(task):
        if task['kind']=='tool_numeric':
            return task['question']+'\nAvailable tools: mean(values: list of numbers), sample_variance(values: list of numbers), dot(a: list of numbers, b: list of numbers). First return only JSON {"tool":"tool_name","args":{...}} with the required arguments. You will receive a tool result next.'
        fmt={'numeric':'a JSON number (not a quoted string); give at least six significant digits if not exact', 'choice':'the single choice letter A, B, C or D', 'rubric':'your concise written response as a JSON string'}[task['kind']]
        return task['question']+'\nReturn only a JSON object with key "answer" containing '+fmt+'.'

    legacy_score=score
    NUM=r'[+-]?(?:\d+(?:\.\d*)?|\.\d+)(?:[eE][+-]?\d+)?'
    def as_number(v):
        if isinstance(v,dict) and set(v)=={'value'}: v=v['value']
        if isinstance(v,bool): return None
        if isinstance(v,(float,int)):
            try: return float(v) if math.isfinite(float(v)) else None
            except OverflowError: return None
        if not isinstance(v,str) or len(v)>100: return None
        v=v.strip()
        try:
            if re.fullmatch(NUM,v): out=float(v)
            elif re.fullmatch(r'[+-]?\d+\s*/\s*[+-]?\d+',v):
                a,b=v.split('/'); out=float(Fraction(int(a),int(b)))
            else: return None
            return out if math.isfinite(out) else None
        except (ValueError,ZeroDivisionError,OverflowError): return None

    def parse_response(raw):
        raw=str(raw).strip()
        if len(raw)>20000: return None,None,False
        def no_dup(pairs):
            d={}
            for k,v in pairs:
                if k in d: raise ValueError('duplicate key')
                d[k]=v
            return d
        def reject(s): raise ValueError('nonfinite')
        wrapped=re.fullmatch(r'```(?:json)?\s*([\s\S]*?)\s*```',raw,re.I)
        text=wrapped.group(1).strip() if wrapped else raw
        # Recover only known presentation artifacts, not arbitrary prose.
        if not wrapped:
            text=re.sub(r'^```(?:json)?\s*','',text,flags=re.I).strip()
            text=re.sub(r'^\|\s*(?=\{)','',text).strip()
        try:
            obj=json.loads(text,object_pairs_hook=no_dup,parse_constant=reject)
            if isinstance(obj,dict): return obj.get('answer') if set(obj)=={'answer'} else None,obj,text==raw
            if isinstance(obj,(str,int,float)) and not isinstance(obj,bool): return obj,None,False
        except (ValueError,RecursionError):
            if re.fullmatch(r'[A-Da-d]|'+NUM+r'|[+-]?\d+\s*/\s*[+-]?\d+',text): return text,None,False
        return None,None,False

    def hardened(t,raw,tool_result=None):
        v,obj,whole=parse_response(raw); k=t['kind']; numeric=as_number(v)
        if k=='rubric':
            protocol=bool(whole and obj and isinstance(v,str) and v.strip())
            return dict(strict_protocol=protocol,answer_parseable=bool(isinstance(v,str) and v.strip()),content_correct=None,operational_correct=None,parsed_value=None)
        if k=='choice':
            match=None
            if isinstance(v,str):
                s=v.strip(); match=s.upper() if re.fullmatch('[A-Da-d]',s) else next((a for a,b in options(t).items() if s.casefold()==b.casefold()),None)
            protocol=bool(whole and obj and isinstance(v,str) and re.fullmatch('[A-D]',v))
            parseable=match is not None; correct=match==t['gold']; val=match
        else:
            protocol=bool(whole and obj and isinstance(v,(int,float)) and not isinstance(v,bool) and numeric is not None)
            parseable=numeric is not None; correct=bool(parseable and numeric_ok(numeric,t['gold'],t['atol'],t['rtol'])); val=numeric
        operational=correct and (k!='tool_numeric' or bool(tool_result and tool_result.get('ok')))
        return dict(strict_protocol=protocol,answer_parseable=parseable,content_correct=int(correct),operational_correct=int(operational),parsed_value=val)

    for t in benchmark:
        if t['kind']=='rubric': continue
        tool=execute_tool(t,canon({'tool':t['tool'],'args':t['tool_args']})) if t['kind']=='tool_numeric' else None
        assert hardened(t,canon({'answer':t['gold']}),tool)['operational_correct']==1
        assert hardened(t,'malformed',tool)['operational_correct']==0
        if t['kind']!='choice':
            assert hardened(t,canon({'answer':{'value':str(t['gold'])}}),tool)['content_correct']==1
            assert not hardened(t,canon({'answer':{'value':str(t['gold'])}}),tool)['strict_protocol']
            assert not hardened(t,'{"answer":true}',tool)['answer_parseable']
    assert as_number('1/4')==0.25 and as_number('1/0') is None
    assert parse_response('{"answer":1,"answer":2}')[0] is None
    print('Scoring self-tests passed for',sum(t['kind']!='rubric' for t in benchmark),'objective tasks')
````

## 3. Automatic run-all controller

````python
# @title Run the complete pipeline — automatically resume and stop at gates
STAGES=['RECOVER','APPROVE','SMOKE','TRAIN','EVALUATE','PACKAGE','PUBLISH']
assert STOP_AFTER in STAGES
CONFIG_NAMES=['USE_DRIVE','INSTALL_GPU_DEPS','PROJECT_ROOT','SOURCE_NAME','D07_FILENAME','FRESH_TEST_PATH','FRESH_TEST_SHA256','FRESH_INDEPENDENCE_ATTESTED','DATA_LICENSE_REVIEWED','DATA_REVIEW_APPROVED','HUMAN_EVAL_APPROVED','PUBLISH_APPROVAL','MAKE_PUBLIC','PUBLIC_APPROVAL','MAX_TRAIN_PER_TEMPLATE','MAX_VAL_PER_TEMPLATE','MAX_LENGTH','GRAD_ACCUM','EPOCHS','SEED','MIN_D07_GAIN','MAX_FAMILY_DROP','MIN_FRESH_GAIN','MAX_PROTOCOL_DROP','BASE','REV','TARGET','EXPECTED_CORPUS']
CONFIG={k:globals()[k] for k in CONFIG_NAMES}
# No credentials are present in CONFIG. The publication worker obtains HF_TOKEN from Colab Secrets only.
assert TARGET=='roime81/trimorious-research-4B-v2'
assert BASE=='Qwen/Qwen3-4B-Instruct-2507' and REV=='cdbee75f17c01a7cc42f958dc650907174af0554'
if FRESH_TEST_SHA256:
    import re
    assert re.fullmatch('[a-f0-9]{64}',FRESH_TEST_SHA256), 'Use the externally sealed lowercase SHA-256.'

def file_sha(p):
    h=hashlib.sha256()
    with Path(p).open('rb') as f:
        for block in iter(lambda:f.read(1024*1024),b''): h.update(block)
    return h.hexdigest()
def read_json(p): return json.loads(Path(p).read_text())
def save_json(p,x):
    p=Path(p); temp=p.with_suffix(p.suffix+'.tmp'); temp.write_text(json.dumps(x,indent=2)); temp.replace(p)
def record_controller(stage,status,detail):
    save_json(WORK/'auto_controller_state.json',{'stage':stage,'status':status,'detail':str(detail)})
def pause(stage,detail):
    record_controller(stage,'waiting',detail); print('\nPAUSED — '+stage+'\n'+str(detail)+'\nSave changes, then Runtime → Run all. Existing artifacts are preserved.',flush=True)
def limited(stage):
    if STAGES.index(stage)>=STAGES.index(STOP_AFTER):
        record_controller(stage,'requested_stop','STOP_AFTER reached'); print('Requested stop after',stage); return True
    return False

def cpu(stage, parts):
    ns={**CONFIG,'USE_DRIVE':False,'STAGE':stage,'__name__':'__main__'}
    for part in ['common']+parts:
        p=SCRIPT_DIR/(part+'.py'); exec(compile(p.read_text(),str(p),'exec'),ns)
    return ns

def gpu(stage):
    print('\n=== '+stage+' — isolated GPU worker ===',flush=True)
    record_controller(stage,'running','GPU subprocess')
    config_path=SCRIPT_DIR/'run_config.json'; save_json(config_path,CONFIG)
    env=os.environ.copy(); env['PYTHONUNBUFFERED']='1'; env.setdefault('PYTORCH_CUDA_ALLOC_CONF','expandable_segments:True')
    p=subprocess.Popen([sys.executable,'-u',str(SCRIPT_DIR/'worker.py'),str(config_path),stage],env=env,stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True,bufsize=1)
    try:
        for line in p.stdout:
            print(line,end='',flush=True)
        status=p.wait()
    except BaseException:
        p.terminate()
        try: p.wait(timeout=20)
        except subprocess.TimeoutExpired: p.kill(); p.wait()
        raise
    if status: raise RuntimeError(f'{stage} worker exited with code {status}. Read the actual error above; checkpoints preserved.')
    # Child exit releases CUDA allocations. Dependencies already installed for subsequent child processes in this session.
    CONFIG['INSTALL_GPU_DEPS']=False

def validate_active(ns):
    approval,run,done,final=ns['active_run']()
    cfg=done['config']
    for k,v in {'base':BASE,'revision':REV,'max_length':MAX_LENGTH,'grad_accum':GRAD_ACCUM,'epochs':EPOCHS,'seed':SEED}.items():
        assert cfg[k]==v,f'Trained adapter uses different {k}; do not silently change a finished run.'
    assert done['steps']>0 and (final/'adapter_model.safetensors').is_file()
    return approval,run,done,final

def get_eval_artifacts(run):
    active=read_json(WORK/'active_evaluation.json'); assert active['run']==run.name
    folder=run/'evaluation'/active['evaluation']; return folder,read_json(folder/'evaluation_report.json')

def eval_receipt(run,done):
    folder,report=get_eval_artifacts(run)
    assert report['evaluation_config']['adapter_hashes']==done['adapter_hashes']
    assert report['evaluation_config']['fresh_hash']==FRESH_TEST_SHA256 and report['fresh']
    db=folder/'responses.sqlite3'
    with sqlite3.connect(db.resolve().as_uri()+'?mode=ro',uri=True) as conn:
        assert conn.execute('PRAGMA integrity_check').fetchone()[0]=='ok'
        counts=dict(conn.execute('SELECT model,COUNT(*) FROM responses GROUP BY model'))
        expected=1000+report['fresh']['n']
        assert counts=={'qwen3_base':expected,'trimorious_v2':expected}, f'Incomplete paired evaluation: {counts}'
    return {'evaluation':folder.name,'report_sha256':file_sha(folder/'evaluation_report.json'),'db_sha256':file_sha(db),'mapping_sha256':file_sha(folder/'rubric_mapping_private.json'),'fresh_sha256':FRESH_TEST_SHA256}

def audit_stage(stage, paths):
    import zipfile
    target=WORK/'stage_audits'; target.mkdir(exist_ok=True)
    files=sorted({Path(p) for p in paths if Path(p).is_file()})
    assert files, 'No audit artifacts for '+stage
    assert all(p.resolve().is_relative_to(WORK.resolve()) for p in files)
    manifest={str(p.relative_to(WORK)):file_sha(p) for p in files}
    digest=hashlib.sha256(json.dumps(manifest,sort_keys=True).encode()).hexdigest()[:20]
    dest=target/f'{stage}_{digest}_PRIVATE.zip'; temp=dest.with_suffix('.tmp')
    if not dest.exists():
        with zipfile.ZipFile(temp,'w',zipfile.ZIP_DEFLATED) as z:
            for p in files: z.write(p,str(p.relative_to(WORK)))
            z.writestr('audit_manifest.json',json.dumps(manifest,indent=2))
        with zipfile.ZipFile(temp) as z:
            assert z.testzip() is None and set(z.namelist())==set(manifest)|{'audit_manifest.json'}
            assert all(hashlib.sha256(z.read(k)).hexdigest()==v for k,v in manifest.items())
        temp.replace(dest)
    with zipfile.ZipFile(dest) as z:
        assert z.testzip() is None and set(z.namelist())==set(manifest)|{'audit_manifest.json'}
        assert json.loads(z.read('audit_manifest.json'))==manifest
        assert all(hashlib.sha256(z.read(k)).hexdigest()==v for k,v in manifest.items())
    print('Private stage audit verified:',dest,flush=True)
    return dest

def main():
    # RECOVER re-verifies the actual source and preserves existing review decisions. No GPU and no new generation.
    print('\n=== RECOVER — verify saved corpus ===',flush=True)
    record_controller('RECOVER','running','Verifying original archive')
    ns=cpu('RECOVER',['recover','approve','eval_setup'])
    audit_stage('RECOVER',[WORK/'recovery.json'])
    if limited('RECOVER'): return
    print('\n=== APPROVE — freeze reviewed data ===',flush=True)
    approval_file=WORK/'dataset_approval.json'
    if approval_file.exists():
        approval=ns['load_approved']()
        assert approval['review_set_sha256']==read_json(WORK/'recovery.json')['review_set_sha256']
        assert not approval['fresh_test_sha256'] or not FRESH_TEST_SHA256 or approval['fresh_test_sha256']==FRESH_TEST_SHA256, 'Previously sealed fresh hash differs.'
        print('Frozen approved dataset verified; reusing.',approval['counts'])
    else:
        if not (DATA_LICENSE_REVIEWED and DATA_REVIEW_APPROVED):
            return pause('APPROVE',f'Review {WORK/"review_queue.csv"}; approve actual accepted rows, then set both data approval flags.')
        ns=cpu('APPROVE',['recover','approve','eval_setup']); approval=ns['load_approved']()
    audit_stage('APPROVE',[WORK/'dataset_approval.json'])
    if limited('APPROVE'): return
    if (WORK/'active_run.json').exists():
        approval,run,done,final=validate_active(ns); print('Verified trained adapter; skipping SMOKE and TRAIN:',run)
        audit_stage('SMOKE',[run/'smoke_pass.json'])
        if STOP_AFTER=='SMOKE': return limited('SMOKE')
    else:
        gpu('SMOKE')
        # Identify the exact matching smoke configuration, not unrelated historical runs.
        smoke_paths=[p for p in (WORK/'sft_output').glob('*/smoke_pass.json') if read_json(p).get('config',{}).get('approval_sha256')==ns['sha'](approval) and all(read_json(p)['config'].get(k)==v for k,v in {'base':BASE,'revision':REV,'max_length':MAX_LENGTH,'grad_accum':GRAD_ACCUM,'epochs':EPOCHS,'seed':SEED}.items())]
        assert smoke_paths, 'Matching smoke receipt missing'
        audit_stage('SMOKE',[*smoke_paths,WORK/'auto_gpu_versions.json'])
        if limited('SMOKE'): return
        gpu('TRAIN')
        approval,run,done,final=validate_active(ns)
    audit_stage('TRAIN',[run/'training_complete.json',run/'config.json'])
    if limited('TRAIN'): return
    # The user can finish training before obtaining an independent test, but no expensive diagnostic-only evaluation is started automatically.
    if not FRESH_TEST_PATH or not FRESH_TEST_SHA256 or not FRESH_INDEPENDENCE_ATTESTED:
        return pause('EVALUATE','Provide externally authored, independently reviewed fresh JSONL, sealed SHA-256 and independence attestation (see instructions). Adapter is saved. Do not generate a test from training examples.')
    fresh_path=Path(FRESH_TEST_PATH)
    assert fresh_path.is_file() and file_sha(fresh_path)==FRESH_TEST_SHA256, 'Fresh confirmation file missing or hash mismatch.'
    d07=ROOT/D07_FILENAME
    if not d07.is_file():
        temp=Path('/content')/D07_FILENAME
        if not temp.is_file(): return pause('EVALUATE',f'Save {D07_FILENAME} in {ROOT}.')
        # Verify before copying to persistent storage; never expose gold or use it for training.
        import zipfile
        with zipfile.ZipFile(temp) as z: ns['archive_check'](z)
        tmp=d07.with_suffix('.zip.tmp'); shutil.copyfile(temp,tmp); assert file_sha(tmp)==file_sha(temp); tmp.replace(d07)
    receipt_path=run/'auto_eval_receipt.json'
    if receipt_path.exists():
        assert read_json(receipt_path)==eval_receipt(run,done), 'Evaluation artifacts changed; investigate before release.'
        print('Verified completed paired evaluation; reusing.')
    else:
        gpu('EVALUATE'); save_json(receipt_path,eval_receipt(run,done))
    folder,report=get_eval_artifacts(run)
    audit_stage('EVALUATE',[folder/'evaluation_report.json',receipt_path])
    print('Objective release checks:',report['checks'])
    if not report['objective_gate_passed']:
        return pause('EVALUATE',f'Release threshold failed. Inspect {folder/"evaluation_report.json"}. Do not lower thresholds or publish.')
    if limited('EVALUATE'): return
    if not (HUMAN_EVAL_APPROVED and DATA_LICENSE_REVIEWED):
        return pause('PACKAGE',f'Review {folder/"rubric_review_private.csv"} and evaluation report; set human evaluation and license flags only after review.')
    print('\n=== PACKAGE — verified release files ===',flush=True)
    package=run/'release_package'; package_receipt=run/'package_approval.json'
    if package_receipt.exists():
        rec=read_json(package_receipt)
        assert rec['manifest_sha256']==file_sha(package/'artifact_manifest.json')
        assert rec['rubric_review_sha256']==file_sha(folder/'rubric_review_private.csv')
        assert rec['evaluation_sha256']==ns['sha'](report) and rec['target']==TARGET
        m=read_json(package/'artifact_manifest.json'); assert all(file_sha(package/k)==v for k,v in m.items())
        assert {p.name for p in package.iterdir()}==set(m)|{'artifact_manifest.json'}
        print('Verified existing release package:',package)
    else:
        cpu('PACKAGE',['approve','eval_setup','package'])
    audit_stage('PACKAGE',[package_receipt,package/'artifact_manifest.json',package/'benchmark_summary.json'])
    if limited('PACKAGE'): return
    if PUBLISH_APPROVAL!='PUBLISH TRIMORIOUS 2.0':
        return pause('PUBLISH',f'Inspect {package}. Add HF_TOKEN to Colab Secrets and explicitly approve private upload.')
    if MAKE_PUBLIC:
        prior=run/'publish_receipt.json'
        if not prior.exists() or not read_json(prior).get('verified'):
            return pause('PUBLISH','First upload privately with MAKE_PUBLIC=False. Inspect the verified private repository before approving public visibility.')
        if PUBLIC_APPROVAL!='MAKE TRIMORIOUS 2.0 PUBLIC': return pause('PUBLISH','Explicit public-visibility approval required.')
    print('\n=== PUBLISH — explicit Hub release ===',flush=True)
    cpu('PUBLISH',['approve','eval_setup','package','publish'])
    receipt=read_json(run/'publish_receipt.json'); assert receipt['verified']
    audit_stage('PUBLISH',[run/'publish_receipt.json',package_receipt])
    if receipt['private']:
        return pause('PUBLISH','Private upload verified. Inspect the repository, then explicitly approve public visibility on a subsequent run.')
    record_controller('PUBLISH','complete','Public repository and remote artifact hashes verified')
    print('TRIMORIOUS 2.0 PUBLIC RELEASE VERIFIED:', 'https://huggingface.co/'+TARGET)

try:
    main()
except KeyboardInterrupt:
    record_controller('INTERRUPTED','interrupted','Stopped by user; rerun with same settings')
    raise
except Exception as exc:
    record_controller('ERROR','blocked',f'{type(exc).__name__}: {exc}')
    raise
````

## 4. Resume and troubleshooting
- **Every restart:** reconnect, select GPU if training/evaluation remain, then Run all. No need to edit a stage variable. All decisions and training/evaluation checkpoints are in `release_v2`. Recovery rechecks the source; it does not regenerate candidates or overwrite the review CSV.
- **GPU OOM or disconnect:** the worker stops; its process is not reused. If needed disconnect/delete the runtime after checking Drive. Keep the same configuration and dependency lock for resume. If changing the training settings, smoke test must match the new config; never mix incompatible checkpoints. If a completed adapter exists, changing its settings is blocked.
- **Incomplete checkpoint:** original trainer refuses an incomplete latest checkpoint. Inspect previous checkpoints; do not delete the only checkpoint or forge completion.
- **Fresh test missing:** training can finish; evaluation pauses. Supply an independently authored set. Do not fabricate test results or auto-approve human review.
- **Publication:** private first; public only on a later explicitly approved run. If creating a repo succeeded before its ownership receipt was saved, inspect manually; the code refuses to overwrite unrelated existing repos.
- **Review artifacts to share:** `recovery.json`, `dataset_approval.json`, `smoke_pass.json`, `training_complete.json`, `evaluation_report.json`, `auto_controller_state.json`. Never share D07 gold or HF_TOKEN.
